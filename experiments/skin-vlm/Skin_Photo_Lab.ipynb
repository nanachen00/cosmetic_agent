# BLINK AI 膚況照片實驗

預設照片為公開授權例圖，只作執行範例。

要測自己的照片，先在左側 data 資料夾上傳，修改下方 IMAGE_PATH，再按 Ctrl＋Enter。上傳原圖會保留在該資料夾；結果不會寫入膚況日記。

此原型只提出外觀候選資訊，未完成膚況準確率或臨床驗證。

```python
from pathlib import Path
import json, subprocess
# 在左側 data 資料夾上傳照片後，修改這個路徑，再按 Ctrl+Enter。
IMAGE_PATH = 'data/public_acne_ccby4.jpg'
NOTE = ''
ROOT = Path('/var/lib/jenkins/tomod/experiments/skin-vlm')
image = (ROOT / IMAGE_PATH).resolve()
result = ROOT / 'results/notebook-latest.json'
proc = subprocess.run([str(ROOT/'.venv/bin/python'),str(ROOT/'skin_vlm.py'),str(image),'--model',str(ROOT/'model'),'--note',NOTE,'--output',str(result)],cwd=ROOT,capture_output=True,text=True,timeout=240)
if proc.returncode: raise RuntimeError(proc.stderr[-1500:])
report = json.loads(result.read_text())
print('狀態：', report['status'])
for item in report.get('observation',{}).get('observations',[]):
    print('•',item['tag'],'/',item['region'],'：',item['evidence'])
for text in report.get('observation',{}).get('limitations',[]): print('限制：',text)
print('補充問題：',report.get('observation',{}).get('follow_up_question',''))
print('推論秒數：',report.get('inference_seconds','品質檢查拒絕'))
print('這是外觀候選資訊，尚未完成準確率驗證；不會自動寫入膚況日記。')
```
